```python
# ============================================================
# RQ1 – CHI-SQUARE TEST OF INDEPENDENCE
# Lead Source × Booking Outcome
# ============================================================

from scipy.stats import chi2_contingency


contingency_table = pd.crosstab(
    df_rq1["Marketing_Measure"],
    df_rq1["IsCancelled"]
)

print("CONTINGENCY TABLE")
print("=" * 60)
print(contingency_table)


chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table
)

print("\nCHI-SQUARE TEST RESULTS")
print("=" * 60)
print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {degrees_of_freedom}")
print(f"P-value: {p_value:.6f}")


alpha = 0.05

print("\nSTATISTICAL DECISION")
print("=" * 60)

if p_value < alpha:
    print("Result: Statistically significant")
    print("Reject the null hypothesis (H₀).")
    print("Lead source and booking outcome are associated.")
else:
    print("Result: Not statistically significant")
    print("Fail to reject the null hypothesis (H₀).")
    print("No statistically significant association was detected.")
```

```python
# ============================================================
# RQ1 – CRAMÉR'S V
# Strength of association between Lead Source and Booking Outcome
# ============================================================

import numpy as np


n = contingency_table.sum().sum()
min_dimension = min(contingency_table.shape) - 1

cramers_v = np.sqrt(
    chi2 / (n * min_dimension)
)

print("CRAMÉR'S V")
print("=" * 60)
print(f"Sample size (N): {n}")
print(f"Cramér's V: {cramers_v:.4f}")

print("\nINTERPRETATION")
print("=" * 60)

if cramers_v < 0.10:
    print("Very weak/negligible association")
elif cramers_v < 0.30:
    print("Weak association")
elif cramers_v < 0.50:
    print("Moderate association")
else:
    print("Strong association")
```

```python
# ============================================================
# RQ1 – STANDARDISED RESIDUALS
# Identify which lead sources drive the association
# ============================================================


expected = pd.DataFrame(
    chi2_contingency(contingency_table)[3],
    index=contingency_table.index,
    columns=contingency_table.columns
)


residuals = (contingency_table - expected) / np.sqrt(expected)

print("EXPECTED FREQUENCIES")
print("=" * 60)
print(expected.round(2))

print("\nSTANDARDISED RESIDUALS")
print("=" * 60)
print(residuals.round(2))

print("\nINTERPRETATION")
print("=" * 60)
print("Residuals above +1.96 indicate more cases than expected.")
print("Residuals below -1.96 indicate fewer cases than expected.")
print("Values between -1.96 and +1.96 are not statistically unusual.")
```

```python
# ============================================================
# RQ1 – CANCELLATION RATE BY LEAD SOURCE
# ============================================================

rq1_rates = (
    df_rq1
    .groupby("Marketing_Measure")["IsCancelled"]
    .agg(
        Total_Events="count",
        Cancelled="sum"
    )
    .reset_index()
)

# Calculate non-cancelled events
rq1_rates["Not_Cancelled"] = (
    rq1_rates["Total_Events"] - rq1_rates["Cancelled"]
)

# Calculate percentages
rq1_rates["Cancellation_Rate_%"] = (
    rq1_rates["Cancelled"] /
    rq1_rates["Total_Events"] * 100
)

rq1_rates["Booking_Rate_%"] = (
    rq1_rates["Not_Cancelled"] /
    rq1_rates["Total_Events"] * 100
)

# Sort from highest to lowest cancellation rate
rq1_rates = rq1_rates.sort_values(
    "Cancellation_Rate_%",
    ascending=False
)

print(rq1_rates.round(2).to_string(index=False))
```

```python
# ============================================================
# RQ1 – 95% CONFIDENCE INTERVALS FOR CANCELLATION RATES
# ============================================================

from statsmodels.stats.proportion import proportion_confint

rq1_ci = rq1_rates.copy()

# Calculate 95% Wilson confidence intervals
ci = rq1_ci.apply(
    lambda row: proportion_confint(
        count=int(row["Cancelled"]),
        nobs=int(row["Total_Events"]),
        alpha=0.05,
        method="wilson"
    ),
    axis=1
)

rq1_ci["CI_Lower_%"] = [x[0] * 100 for x in ci]
rq1_ci["CI_Upper_%"] = [x[1] * 100 for x in ci]

print(
    rq1_ci[
        [
            "Marketing_Measure",
            "Total_Events",
            "Cancelled",
            "Cancellation_Rate_%",
            "CI_Lower_%",
            "CI_Upper_%"
        ]
    ].round(2).to_string(index=False)
)
```

```python
# ============================================================
# RQ2 – CREATE RESPONDENT × DECISION-FACTOR MATRIX
# ============================================================

# Select the decision-factor responses
df_rq2_multi = df_multiselect[
    df_multiselect["QuestionKey"] == "decision_factors"
].copy()

# Create binary respondent × factor matrix
rq2_matrix = pd.crosstab(
    df_rq2_multi["ResponseID"],
    df_rq2_multi["Answer"]
)

# Convert to binary: selected = 1, not selected = 0
rq2_matrix = (rq2_matrix > 0).astype(int)

print("RQ2 MATRIX")
print("=" * 60)
print("Respondents:", rq2_matrix.shape[0])
print("Factors:", rq2_matrix.shape[1])

print("\nNumber of respondents selecting each factor:")
print(rq2_matrix.sum().sort_values(ascending=False))

print("\nFirst 5 respondents:")
print(rq2_matrix.head())
```

```python
# ============================================================
# RQ2 – CHECK FOR MISSING VALUES
# ============================================================

print("Missing values in RQ2 matrix:")
print(rq2_matrix.isna().sum())

print("\nTotal missing values:", rq2_matrix.isna().sum().sum())

print("\nData type of variables:")
print(rq2_matrix.dtypes)
```

```python
# ============================================================
# RQ2 – COCHRAN'S Q TEST
# Venue Selection Factors
# ============================================================

# Install Pingouin if it is not already installed
!pip -q install pingouin

import pandas as pd
import numpy as np
import pingouin as pg

# ------------------------------------------------------------
# 1. Check the RQ2 matrix
# ------------------------------------------------------------

print("RQ2 DATA CHECK")
print("=" * 60)

print(f"Number of respondents: {rq2_matrix.shape[0]}")
print(f"Number of venue-selection factors: {rq2_matrix.shape[1]}")
print(f"Total missing values: {rq2_matrix.isna().sum().sum()}")

print("\nSelection counts:")
print(rq2_matrix.sum().sort_values(ascending=False))


# ------------------------------------------------------------
# 2. Run Cochran's Q test
# ------------------------------------------------------------

cochran_result = pg.cochran(
    data=rq2_matrix
)

print("\n" + "=" * 60)
print("COCHRAN'S Q TEST")
print("=" * 60)

print(cochran_result)


# ------------------------------------------------------------
# 3. Extract results
# ------------------------------------------------------------

Q = cochran_result["Q"].iloc[0]
p = cochran_result["p_unc"].iloc[0]
df = cochran_result["dof"].iloc[0]


# ------------------------------------------------------------
# 4. Print clean statistical result
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STATISTICAL RESULT")
print("=" * 60)

print(f"Cochran's Q = {Q:.3f}")
print(f"Degrees of freedom = {int(df)}")
print(f"P-value = {p:.3e}")

if p < 0.05:
    print("\nResult: STATISTICALLY SIGNIFICANT")
    print("Decision: Reject H₀.")
    print(
        "Conclusion: The nine venue-selection factors "
        "were not selected at equal rates."
    )
else:
    print("\nResult: NOT STATISTICALLY SIGNIFICANT")
    print("Decision: Fail to reject H₀.")
    print(
        "Conclusion: No statistically significant difference "
        "was detected between the selection rates."
    )


# ------------------------------------------------------------
# 5. Display selection percentages
# ------------------------------------------------------------

selection_summary = pd.DataFrame({
    "Factor": rq2_matrix.columns,
    "Selected": rq2_matrix.sum().values,
    "Not_Selected": (1 - rq2_matrix).sum().values
})

selection_summary["Selection_Rate_%"] = (
    selection_summary["Selected"]
    / len(rq2_matrix)
    * 100
)

selection_summary = selection_summary.sort_values(
    "Selection_Rate_%",
    ascending=False
)

print("\n" + "=" * 60)
print("VENUE-SELECTION FACTOR RANKING")
print("=" * 60)

print(
    selection_summary.round(2).to_string(index=False)
)
```

```python
# ============================================================
# RQ2 – POST-HOC MCNEMAR TESTS
# Pairwise comparison of venue-selection factors
# Bonferroni correction applied
# ============================================================

from itertools import combinations
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.multitest import multipletests

# Store results
posthoc_results = []

# Get all factors
factors = rq2_matrix.columns.tolist()

# Number of pairwise comparisons
num_comparisons = len(list(combinations(factors, 2)))

print("Number of pairwise comparisons:", num_comparisons)

# ------------------------------------------------------------
# Run pairwise McNemar tests
# ------------------------------------------------------------

for factor1, factor2 in combinations(factors, 2):

    # 2 × 2 table:
    #             Factor 2
    #             0      1
    # Factor 1
    # 0
    # 1
    table = pd.crosstab(
        rq2_matrix[factor1],
        rq2_matrix[factor2]
    )

    # Ensure 2 × 2 structure
    table = table.reindex(
        index=[0, 1],
        columns=[0, 1],
        fill_value=0
    )

    # McNemar test
    result = mcnemar(
        table,
        exact=True
    )

    posthoc_results.append({
        "Factor_1": factor1,
        "Factor_2": factor2,
        "Chi2": result.statistic,
        "P_value": result.pvalue
    })

# Convert results to DataFrame
posthoc_df = pd.DataFrame(posthoc_results)

# ------------------------------------------------------------
# Bonferroni correction
# ------------------------------------------------------------

reject, p_adjusted, _, _ = multipletests(
    posthoc_df["P_value"],
    alpha=0.05,
    method="bonferroni"
)

posthoc_df["Adjusted_P"] = p_adjusted
posthoc_df["Significant"] = reject

# Sort by adjusted p-value
posthoc_df = posthoc_df.sort_values(
    "Adjusted_P"
)

# ------------------------------------------------------------
# Display all results
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("PAIRWISE MCNEMAR TESTS WITH BONFERRONI CORRECTION")
print("=" * 80)

print(
    posthoc_df[
        [
            "Factor_1",
            "Factor_2",
            "Chi2",
            "P_value",
            "Adjusted_P",
            "Significant"
        ]
    ].round(4).to_string(index=False)
)

# ------------------------------------------------------------
# Display only statistically significant comparisons
# ------------------------------------------------------------

significant_pairs = posthoc_df[
    posthoc_df["Significant"] == True
]

print("\n" + "=" * 80)
print("STATISTICALLY SIGNIFICANT PAIRWISE DIFFERENCES")
print("=" * 80)

if len(significant_pairs) > 0:
    print(
        significant_pairs[
            [
                "Factor_1",
                "Factor_2",
                "Adjusted_P"
            ]
        ].round(4).to_string(index=False)
    )
else:
    print("No statistically significant pairwise differences were found.")
```

```python
# ============================================================
# RQ3 – LEAD-SOURCE STRATEGIC PERFORMANCE CLASSIFICATION
# ============================================================

# Start from the CRM summary already created
strategic_sources = crm_summary.copy()

# Calculate median thresholds
volume_median = strategic_sources["Total_Events"].median()
booking_median = strategic_sources["Booking_Rate_%"].median()

print("=" * 80)
print("RQ3 – STRATEGIC LEAD-SOURCE ANALYSIS")
print("=" * 80)

print(f"Median event volume: {volume_median:.1f}")
print(f"Median booking rate: {booking_median:.2f}%")

# ------------------------------------------------------------
# Classify each lead source
# ------------------------------------------------------------

def classify_source(row):
    high_volume = row["Total_Events"] >= volume_median
    high_booking = row["Booking_Rate_%"] >= booking_median

    if high_volume and high_booking:
        return "High Volume / High Booking Rate"
    elif high_volume and not high_booking:
        return "High Volume / Low Booking Rate"
    elif not high_volume and high_booking:
        return "Low Volume / High Booking Rate"
    else:
        return "Low Volume / Low Booking Rate"


strategic_sources["Strategic_Category"] = (
    strategic_sources.apply(classify_source, axis=1)
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

strategic_sources = strategic_sources.sort_values(
    "Booking_Rate_%",
    ascending=False
)

print("\nLEAD-SOURCE STRATEGIC CLASSIFICATION")
print("-" * 80)

print(
    strategic_sources[
        [
            "Marketing_Measure",
            "Total_Events",
            "Booking_Rate_%",
            "Cancellation_Rate_%",
            "Event_Share_%",
            "Strategic_Category"
        ]
    ].round(2).to_string(index=False)
)
```
